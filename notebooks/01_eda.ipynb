# 01 — Exploratory Data Analysis
EDA sử dụng các module có sẵn trong `analysis_functions`; notebook chỉ chọn dữ liệu và điều phối biểu đồ.

```python
from pathlib import Path
import sys, json, pandas as pd
from IPython.display import Image, display
ROOT = Path.cwd().resolve(); ROOT = ROOT.parent if ROOT.name == 'notebooks' else ROOT
sys.path[:0] = [str(ROOT / 'src'), str(ROOT)]
from cooling_load.config import load_config
from cooling_load.io import read_frame
from cooling_load.pipeline import run_eda
from analysis_functions.basic_data_inspection import DataInspector, DataTypesInspectionStrategy, SummaryStatisticsInspectionStrategy
config = load_config(ROOT / 'configs/base.yaml')
raw = read_frame(config.path('raw_data'))
raw['timestamp'] = pd.to_datetime(raw['timestamp'], utc=True)
```

```python
DataInspector(DataTypesInspectionStrategy()).execute_inspection(raw)
DataInspector(SummaryStatisticsInspectionStrategy()).execute_inspection(raw)
display(raw.isna().mean().sort_values(ascending=False).to_frame('missing_fraction'))
```

```python
eda_report = run_eda(config, raw)
print('Summary:', eda_report.summary_path)
print('Saved figures:')
for figure in eda_report.figures:
    print(' -', figure)
```

```python
display(json.loads(eda_report.summary_path.read_text()))
for figure in eda_report.figures:
    display(Image(filename=str(figure), width=900))
```
