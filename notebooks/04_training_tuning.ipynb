```python
from pathlib import Path
import sys, matplotlib.pyplot as plt, seaborn as sns
ROOT = Path.cwd().resolve(); ROOT = ROOT.parent if ROOT.name == 'notebooks' else ROOT
sys.path[:0] = [str(ROOT / 'src'), str(ROOT)]
from cooling_load.config import load_config
from cooling_load.io import read_frame
from cooling_load.pipeline import run_training
from cooling_load.artifacts import load_bundle
from cooling_load.modeling import tree_feature_importance
config = load_config(ROOT / 'configs/base.yaml')
featured = read_frame(config.path('feature_data'))
result = run_training(config, featured)
```

```python
display(result['leaderboard'])
display(result['tuning_results'])
print('Validation:', result['validation_metrics'])
print('Test:', result['test_metrics'])
print('Local artifact:', result['model_path'])
print('S3 publication:', result['published_model'])
```

```python
bundle = load_bundle(config.path('model'))
importance = tree_feature_importance(bundle.model, top_n=20)
display(importance)
plt.figure(figsize=(9,7)); sns.barplot(data=importance, x='importance', y='feature'); plt.title('Tree feature importance'); plt.show()
```
