Deduplicate, resample theo giờ, forward-fill sensor trong giới hạn và giữ target thiếu để tránh tạo ground truth giả.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve(); ROOT = ROOT.parent if ROOT.name == 'notebooks' else ROOT
sys.path[:0] = [str(ROOT / 'src'), str(ROOT)]
from cooling_load.config import load_config
from cooling_load.io import read_frame
from cooling_load.pipeline import run_preprocessing
config = load_config(ROOT / 'configs/base.yaml')
raw = read_frame(config.path('raw_data'))
clean = run_preprocessing(config, raw)
```

```python
quality = {'raw_rows': len(raw), 'clean_rows': len(clean), 'duplicate_keys_after': int(clean.duplicated(['building_id','timestamp']).sum()), 'target_missing': int(clean.cooling_load_kwh.isna().sum())}
quality
```
